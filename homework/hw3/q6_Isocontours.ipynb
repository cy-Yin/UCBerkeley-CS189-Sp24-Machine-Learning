```python
import numpy as np
import scipy
import matplotlib.pyplot as plt
import os
```

```python
def cal_PDF_grid(mu, cov, x_axis, y_axis):
    '''
    Calculate the probability density function with the given normal distribution.

    Args:
        mu (n-vector): the mean of random variable
        cov (matrix): the variance of random variable
        x_axis (np.array): the x-axis range of the to-be-calculated PDF
        y_axis (np.array): the y-axis range of the to-be-calculated PDF

    Returns:
        matrix: the pdf at the 2D grid with range (x_axis, y_axis)
    '''
    # Create a 2D grid 
    X, Y = np.meshgrid(x_axis, y_axis)
    # Stack the grid points into position vectors for PDF evaluation
    pos = np.dstack((X, Y))
    # Create a multivariate normal distribution object
    rv = scipy.stats.multivariate_normal(mu, cov)
    # Calculate the probability density function (PDF) at each grid point
    Z = rv.pdf(pos)

    return Z
```

```python
def plot_isocontours(x_axis, y_axis, pdf_grid, fig_name):
    '''
    Plot the isocontours with the given probability density function.

    Args:
        x_axis (np.array): the x-axis range of the to-be-plot figure
        y_axis (np.array): the y-axis range of the to-be-plot figure
        pdf_grid (matrix): the PDF of the given random variable at the 2D grid
        fig_name (string): the name of the to-be-saved figure
    '''
    # Create a 2D grid
    X, Y = np.meshgrid(x_axis, y_axis)
    Z = pdf_grid

    # Plot contour lines for the PDF
    plt.figure()
    contour = plt.contour(X, Y, Z, levels=10, cmap='viridis')
    plt.clabel(contour, inline=1, fontsize=10)
    plt.title('Isocontours of 2D Normal Distribution')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.axis('equal')
    
    folder = 'q6_figs'
    if not os.path.exists(folder):
        os.makedirs(folder)
    plt.savefig(os.path.join(folder, fig_name))
    plt.show()
```

```python
# q6.1

mu = np.array([1, 1])
cov = np.array([[1, 0], [0, 2]])
x1 = np.linspace(-2, 4, 500)
y1 = np.linspace(-2, 4, 500)
pdf_grid1 = cal_PDF_grid(mu, cov, x1, y1)
plot_isocontours(x1, y1, pdf_grid1, "q6_1.png")
```

```python
# q6.2

mu = np.array([-1, 2])
cov = np.array([[2, 1], [1, 4]])
x2 = np.linspace(-5, 3, 500)
y2 = np.linspace(-3, 7, 500)
pdf_grid2 = cal_PDF_grid(mu, cov, x2, y2)
plot_isocontours(x2, y2, pdf_grid2, "q6_2.png")
```

```python
# q6.3

mu1 = np.array([0, 2])
cov1 = np.array([[2, 1], [1, 1]])
mu2 = np.array([2, 0])
cov2 = np.array([[2, 1], [1, 1]])
x3 = np.linspace(-3, 5, 500)
y3 = np.linspace(-2, 4, 500)
pdf_grid1 = cal_PDF_grid(mu1, cov1, x3, y3)
pdf_grid2 = cal_PDF_grid(mu2, cov2, x3, y3)
pdf_grid = pdf_grid1 - pdf_grid2
plot_isocontours(x3, y3, pdf_grid, "q6_3.png")
```
