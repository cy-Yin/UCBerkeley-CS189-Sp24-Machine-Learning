```python
import numpy as np
import matplotlib.pyplot as plt
import os
np.random.seed(42)
```

```python
# generate (X1, X2) random sample points
x1 = np.random.normal(loc=3, scale=3, size=100)
x2_noise = np.random.normal(loc=4, scale=2, size=100)
x2 = 0.5 * x1 + x2_noise
sample = np.vstack((x1, x2)) # shape = 2 * 100
```

```python
# q7.1 Compute the mean of the sample
mean = np.mean(sample, axis=1)
print('The mean of the sample is {0}'.format(mean))
```

```python
# q7.2 Compute the covariance matrix
cov = np.cov(sample)
print("The covariance matrix of the sample is")
print(cov)
```

```python
# q7.3 Compute the eigenvalues/eigenvectors of the covariance matrix
eigenvalues, eigenvectors = np.linalg.eig(cov)
print("Eigenvalues:\n", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

```python
# q7.4 Plot the samples

x = sample[0]
y = sample[1]
plt.scatter(x, y, alpha=0.7, s=15) # plot all the 100 sample data points
# plot the eigenvectors
for i in range(len(eigenvalues)):
    vec = eigenvectors[:, i]
    # magnitude = corresponding eigenvalues
    plt.arrow(mean[0], mean[1],
              vec[0] * eigenvalues[i], vec[1] * eigenvalues[i],
              color='r', width=0.1, head_width=0.5, length_includes_head=True)
plt.xlim(-15, 15)
plt.ylim(-15, 15)
plt.xlabel(r"$X_1$")
plt.ylabel(r"$X_2$")
plt.gca().set_aspect('equal', adjustable='box')
plt.title("Sample Points with Covariance Eigenvectors")
folder = 'q7_figs'
if not os.path.exists(folder):
    os.makedirs(folder)
plt.savefig(os.path.join(folder, "q7_4.png"))
plt.show()
```

```python
# Center and rotate the sample points

sample_centered = sample - mean[:, np.newaxis]

# Eigenvectors have been normalized guaranteed by np.linalg.eig() func.
# Therefore U = [v1, v2] is exactly the eigenvectors matrix.
sample_centered_rotated = np.dot(eigenvectors.T, sample_centered)

x_rotated = sample_centered_rotated[0]
y_rotated = sample_centered_rotated[1]
plt.scatter(x_rotated, y_rotated, alpha=0.7, s=15) # plot all the 100 sample data points
# plot the eigenvectors
rotated_eigenvetors = np.array([[1, 0], [0, 1]])
for i in range(len(eigenvalues)):
    vec = rotated_eigenvetors[:, i]
    # magnitude = corresponding eigenvalues
    plt.arrow(0, 0,
              vec[0] * eigenvalues[i], vec[1] * eigenvalues[i],
              color='r', width=0.1, head_width=0.5, length_includes_head=True)
plt.xlim(-15, 15)
plt.ylim(-15, 15)
plt.xlabel(r"$X_1$")
plt.ylabel(r"$X_2$")
plt.gca().set_aspect('equal', adjustable='box')
plt.title("Centered and Rotated Sample Points")
folder = 'q7_figs'
if not os.path.exists(folder):
    os.makedirs(folder)
plt.savefig(os.path.join(folder, "q7_5.png"))
plt.show()
```
